```python
import numpy as np
import time
import pyvisa

rm = pyvisa.ResourceManager()
print(rm.list_resources())
awg = rm.open_resource("USB0::0x1AB1::0x0646::DG9Q26BM00058::INSTR")

awg.timeout = 5000
```

```python
awg.write("*CLS")
#awg.write("SOUR1:FUNC ARB")
awg.write("SOUR1:APPL:DC 0,0,3.0,0")
awg.write("OUTP1 ON")
print(awg.query("SYST:ERR?"))
```

```python
print(awg.query("SOUR1:APPL?"))
print(awg.query("SOUR1:VOLT:OFFS?"))
print(awg.query("SYST:ERR?"))
```
